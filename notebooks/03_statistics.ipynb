## Interpretation

All five tests reject H0 at α = 0.05 on this dataset:

1. **Rating differs across categories** (Kruskal-Wallis) — different product
   categories have different typical ratings.
2. **Discount % differs across categories** (ANOVA) — discounting strategy
   clearly varies by category (e.g. Office Products discount far less than
   Home Improvement).
3. **Price vs. rating** (Spearman) — a very weak but statistically significant
   *positive* association; price alone is a poor predictor of rating.
4. **Discount vs. popularity** (Pearson) — a weak, statistically significant
   *negative* association; heavier discounting is not associated with higher
   popularity scores here.
5. **Review length by rating group** (Mann-Whitney) — review length
   distributions differ significantly between higher- and lower-rated products.

Caveat: with n≈1,300 rows, even small effect sizes (e.g. rho≈0.08) can reach
statistical significance — statistical significance here should not be read as
*practical* significance without also considering the effect size.
