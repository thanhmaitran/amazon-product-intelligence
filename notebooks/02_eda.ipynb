## C. Pricing Analysis

```python
from src.eda import section_pricing

pricing_insights = section_pricing(df)
pricing_insights
```

**Insight:** Price is heavily right-skewed (skew ≈ 4.3) — most products are

## E. Discount Analysis

```python
from src.eda import section_discount

discount_insights = section_discount(df)
print("Discount~Rating correlation:", round(discount_insights["discount_rating_corr"], 3))
discount_insights["biggest_discount_categories"]
```

**Insight:** Discount and rating are weakly *negatively* correlated — heavily
discounted products do not receive systematically better ratings, and in this
dataset the relationship trends slightly the other way. See `03_statistics.ipynb`
for the formal significance test.
