## TF-IDF Keywords per Category

## Topic Modelling (NMF, 6 topics)

```python
from src.nlp import topic_model

topics = topic_model(df, n_topics=6)
for topic, words in topics.items():
    print(f"{topic}: {', '.join(words)}")
```

**Insight:** The topics recovered are coherent and map cleanly onto product
sub-domains without any manual labelling: charging cables (Topic 2), phones/
cameras (Topic 3), TVs (Topic 4), smartwatches (Topic 5), and audio/peripherals
(Topic 6) — validating that the underlying text cleaning (Phase 1) produced
usable, non-noisy tokens. An earlier pass surfaced a topic dominated by leaked
image-URL fragments (`https`, `jpg`, `webp`, ...); this was traced back to a
formatting issue in the raw `review_content` field and fixed by adding URL
stripping to `src/utils.basic_clean` — a good example of why Phase 4 EDA on
text is itself a data-quality check on Phase 1.
